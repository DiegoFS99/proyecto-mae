# MAE Competencia Clasificacion 202620

El gerente de UniBank está interesado en promocionar un producto de reciente desarrollo, el depósito bancario, para lo cual resulta esencial identificar los clientes que tendrían una mayor probabilidad de adquirirlo una vez sea lanzado al mercado. Para el efecto tiene a su disposición la siguiente información relacionada con la última campaña promocional de un producto similar realizada por el banco: los clientes actuales del banco (edad, tipo de trabajo, nivel educativo, entre otras); la forma en que se realizó el contacto durante la campaña (duración de la llamada, número de llamadas que atendió, entre otras); y la decisión finalmente adoptada (adquirió o no el producto).

Dataset Description
Dataset Description Variables de entrada:

'# - índice que indica la posición de la observación.

Datos del cliente:

1 - Age

2 - Job: tipo de trabajo (categórico: "admin.", "Desconocido", "desempleado", "gerencia", "empleada doméstica", "emprendedor", "estudiante", "obrero", "autónomo" , "jubilado", "técnico", "servicios")

3 – Marital Status: estado civil (categórico: "casado", "divorciado", "soltero"; nota: "divorciado" significa divorciado o viudo)

4 - Education (categórica: "desconocida", "secundaria", "primaria", "terciaria")

5 - Credit: ¿tiene un crédito con incumplimiento? (binario: "sí", "no")

6 – Balance.euros: saldo medio anual, en euros

7 – Housing.Loan: ¿tiene préstamo para vivienda? (binario: "sí", "no")

8 – Personal.Loan: ¿tiene préstamo personal? (binario: "sí", "no")

Relacionado con el último contacto:

9 - Contact: tipo de comunicación de contacto (categórico: "desconocido", "teléfono", "celular")

10 – Last.Contact.Day: último día de contacto del mes.

11 - Last.Contact.Month: último mes de contacto del año (categórico: "jan", "feb", "mar",…, "nov", "dec")

12 - Last.Contact.Duration: duración del último contacto, en segundos.

Otros:

13 - Campaign: número de contactos realizados durante esta campaña y para este cliente (incluye el último contacto)

14 - pdays: número de días que pasaron después de que el cliente fue contactado por última vez desde una campaña anterior (-1 significa que el cliente no fue contactado previamente)

15 - Previous número de contactos realizados antes de esta campaña y para este cliente.

16 - poutcome: resultado de la campaña de marketing anterior (categórico: "desconocido", "otro", "fracaso", "éxito")

17 -¿el cliente adquirió el producto? (binario: "sí", "no")

```python
import kagglehub

path = kagglehub.competition_download("mae-competencia-clasificacion-202620")
data_dir = Path(path)
print("Carpeta de datos:", data_dir)
print("Archivos disponibles:")
for archivo in sorted(data_dir.rglob("*")):
    if archivo.is_file():
        print(" -", archivo.relative_to(data_dir))
```

```python
import pandas as pd

# Usa la carpeta obtenida en la celda de descarga.
df_train = pd.read_csv(data_dir / "Train bank.csv")
df_test = pd.read_csv(data_dir / "Test bank.csv")

print("Entrenamiento (filas, columnas):", df_train.shape)
display(df_train.head())
print("Prueba (filas, columnas):", df_test.shape)
display(df_test.head())
```

```python
# listado de variables categoricas con listado de sus valores unicos
valores_unicos = {
    col: sorted(df_train[col].dropna().unique())
    for col in df_train.select_dtypes(include='object').columns
}
valores_unicos # utilio la extencion Data Wrangler para ver estos valores unicos de manera mas visual
```

```python
# Para vitar interpretar -1 como días transcurridos, lo remplazamos por un nan
# Previous, cuyo valor 0 siempre que Pdays es -1 e identifica la ausencia de contactos previos.
df_train["Pdays"] = (
    df_train["Pdays"].replace(-1, float("nan"))
)

# Variables disponibles antes de la campana y variable objetivo.
columnas_seleccionadas = [
    "Age", "Job", "Education", "Marital.Status",
    "Balance..euros.", "Credit", "Housing.Loan", "Personal.Loan",
    "Previous", "Pdays", "Poutcome", "Subscription",
]
df_train = df_train.loc[:, columnas_seleccionadas].copy()

# Convertir yes/no a 1/0
columnas_binarias = ["Credit", "Housing.Loan", "Personal.Loan"]
for columna in columnas_binarias:
    df_train[columna] = df_train[columna].replace({"yes": 1, "no": 0}).astype("int64")
```

```python
# Excluir variables binarias, incluida la variable objetivo
binarias = ["Credit", "Housing.Loan", "Personal.Loan", "Subscription"]
datos_numericos = df_train.select_dtypes(include="number").drop(
    columns=binarias, errors="ignore"
)

media = datos_numericos.mean()
desviacion = datos_numericos.std(ddof=1)

medidas_descriptivas = pd.DataFrame({
    "N validos": datos_numericos.count(), # por los casos nan de Pdays, para identificar la cantidad de observaciones
    "Moda": pd.Series({col: datos_numericos[col].mode().tolist() for col in datos_numericos}),
    "Mediana": datos_numericos.median(),
    "Media": media,
    "Minimo": datos_numericos.min(),
    "Maximo": datos_numericos.max(),
    "Primer cuartil (Q1)": datos_numericos.quantile(0.25),
    "Tercer cuartil (Q3)": datos_numericos.quantile(0.75),
    "Rango": datos_numericos.max() - datos_numericos.min(),
    "Rango intercuartil": datos_numericos.quantile(0.75) - datos_numericos.quantile(0.25),
    "Varianza": datos_numericos.var(ddof=1),
    "Desviacion estandar": desviacion,
    "CV (%)": desviacion / media.replace(0, float("nan")) * 100,
})
medidas_descriptivas.index.name = "Variable"

medidas_descriptivas.round(2)
```
